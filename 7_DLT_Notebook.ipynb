```python
looktables_rule = {
    "rule1" : "show_id is NOT NULL"
}
```

```python
@dlt.table(
    name = "gold_netflixdirectors"
)
@dlt.expect_all_or_drop(looktables_rule)
def myfunc():
    df = spark.readStream.format("delta").load("abfss://silver@netflixprojectdlrayyan.dfs.core.windows.net/netflix_directors")
    return df
```
